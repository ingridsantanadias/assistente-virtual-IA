```python
# Instalação da biblioteca oficial do Gemini
!pip install -U google-genai

from google.colab import userdata
import os

# Configurando a chave de API (salva nos Segredos do Colab com o nome GEMINI_API_KEY)
os.environ["GEMINI_API_KEY"] = userdata.get('GEMINI_API_KEY')

from google import genai
client = genai.Client()
```

```python
base_de_conhecimento = """
[Tópico 1: Processamento de Grandes Volumes de Dados]
- Pergunta Comum: Qual ferramenta usar para dados em larga escala?
- Solução Recomendada: PySpark / Apache Spark rodando em containers Docker para garantir reprodutibilidade e escalabilidade.

[Tópico 2: Automação e Painéis Rápidos]
- Pergunta Comum: Como apresentar indicadores de negócio rapidamente?
- Solução Recomendada: Streamlit combinado com Pandas e Power BI para visualização de dados dinâmica e interativa.

[Tópico 3: Monitoramento IoT]
- Pergunta Comum: Como armazenar dados de sensores de turbidez de água?
- Solução Recomendada: Utilizar microcontroladores como ESP32 integrados à plataforma ThingSpeak para armazenamento e coleta em tempo real.
"""
```

```python
system_instruction = f"""
Você é um assistente virtual especialista em orientar estudantes e profissionais de tecnologia em projetos de Análise de Dados, Engenharia de Dados e IoT.
Sua missão é responder às dúvidas da pessoa usuária utilizando **estritamente** as informações contidas na Base de Conhecimento abaixo.

Regras importantes:
1. Se a resposta não estiver na Base de Conhecimento, diga educadamente: "Desculpe, não possuo informações suficientes sobre isso na minha base de conhecimento atual." e sugira o próximo passo.
2. Seja claro, objetivo, didático e mantenha um tom profissional.

Base de Conhecimento:
{base_de_conhecimento}
"""
```

```python
def perguntar_ao_assistente(pergunta_usuario):
    response = client.models.generate_content(
    model='gemini-3.5-flash-lite',
    contents=pergunta_usuario,
    config={
        'system_instruction': system_instruction,
        'temperature': 0.2,
    }
)

    return response.text

# Testando o assistente interativamente
pergunta = "Qual ferramenta devo usar para dados em larga escala?"
print(f"Usuário: {pergunta}")
print(f"Assistente:\n{perguntar_ao_assistente(pergunta)}")
```
